```python
from transformers import AutoConfig

MODEL_NAME = "facebook/wav2vec2-xls-r-300m"

config = AutoConfig.from_pretrained(MODEL_NAME)

print("=" * 55)
print("XLS-R-300M ARCHITECTURE")
print("=" * 55)

print("Model type:", config.model_type)
print("Encoder layers:", config.num_hidden_layers)
print("Hidden size:", config.hidden_size)
print("Attention heads:", config.num_attention_heads)
print("Intermediate size:", config.intermediate_size)
print("Feature extractor:", config.conv_dim)
print("Feature strides:", config.conv_stride)
```

```python
from transformers import Wav2Vec2ForCTC
import torch

VOCAB_SIZE = 36

base_model = Wav2Vec2ForCTC.from_pretrained(
    MODEL_NAME,
    vocab_size=VOCAB_SIZE,
    pad_token_id=35
)

print("=" * 55)
print("ENCODER ATTENTION MODULES")
print("=" * 55)

for name, module in base_model.named_modules():

    if any(
        target in name
        for target in [
            "q_proj",
            "k_proj",
            "v_proj",
            "out_proj"
        ]
    ):
        print(name)
```

```python
from pathlib import Path
import json

PROJECT_DIR=Path.cwd()

CHECKPOINT_DIR=PROJECT_DIR/"checkpoints"
MODEL_DIR=PROJECT_DIR/"models"

print("="*55)
print("Available Checkpoints")
print("="*55)

for path in sorted(CHECKPOINT_DIR.iterdir()):
    print(path.name)

print("\n"+"="*55)
print("SAVED ADAPTERS")
print("="*55)

for config_path in MODEL_DIR.rglob("adapter_config.json"):

    print("\nAdapter:",config_path.parent)

    with open(config_path,"r") as f:
        config=json.load(f)

    print("Rank:",config.get("r"))
    print("Alpha:",config.get("lora_alpha"))
    print("Target modules:",config.get("target_modules"))
    print("Modules to save:",config.get("modules_to_save"))
```
